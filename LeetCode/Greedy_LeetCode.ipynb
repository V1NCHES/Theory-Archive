# Жадные алгоритмы (Greedy Algorithms)

**Жадный алгоритм** — это метод решения задач, основанный на принятии локально оптимального решения на каждом шаге в надежде, что это приведет к глобально оптимальному решению.

### Основные принципы:
1.  **Жадный выбор**: На каждом шаге мы выбираем лучший вариант «здесь и сейчас», не задумываясь о будущих последствиях.
2.  **Оптимальная подструктура**: Оптимальное решение всей задачи строится из оптимальных решений её подзадач.

### Когда жадность работает?
Жадные алгоритмы работают не всегда. Для их применения нужно доказать две вещи:
*   **Принцип жадного выбора**: Локально лучший выбор не закроет путь к глобальному оптимуму.
*   **Оптимальность подзадач**: Решив подзадачу жадно, мы упростим исходную задачу до такой же, но меньшего размера.

### Типичные стратегии:
*   **Сортировка**: Часто жадное решение начинается с сортировки данных по какому-то признаку (цене, весу, времени).
*   **Подсчет частот**: Выбор наиболее или наименее часто встречающихся элементов.

### 409. Самый длинный палиндром https://leetcode.com/problems/longest-palindrome/

Если задана строка `s`, состоящая из строчных или заглавных букв, верните длину самого длинного палиндрома.

**Теория для задачи**: Палиндром строится из пар одинаковых букв. Если буква встречается 5 раз, мы можем жадно взять 4 из них (2 пары) для палиндрома. Если останутся лишние буквы, одну из них можно поставить в центр.

```python
from collections import Counter

class Solution:
    def longestPalindrome(self, s: str) -> int:
        # 1. Считаем сколько раз встречается каждая буква
        counts = Counter(s)
        length = 0
        has_odd = False
        
        # 2. Жадный выбор: забираем все пары из каждой буквы
        for count in counts.values():
            # Если букв четное количество, берем все
            # Если нечетное (например 3), берем (3 - 1) = 2
            length += (count // 2) * 2
            
            # Если осталась хотя бы одна лишняя буква, помечаем это
            if count % 2 == 1:
                has_odd = True
        
        # 3. Если есть хоть одна лишняя буква, ставим её в центр палиндрома (+1 к длине)
        return length + 1 if has_odd else length
```

**Пояснение**: 
Мы действуем жадно, так как на каждом шаге (для каждой буквы) мы максимизируем её вклад в длину палиндрома, забирая все возможные пары. Нам не нужно знать, какие буквы мы возьмем дальше — пары всегда выгодны.

### 455. Назначение файлов cookie https://leetcode.com/problems/assign-cookies/

У каждого ребенка есть коэффициент жадности `g[i]`, а у каждого печенья — размер `s[j]`. Цель — максимизировать количество довольных детей.

**Теория для задачи**: Это классическая жадная задача. Чтобы удовлетворить максимум детей, нужно тратить ресурсы (печенье) максимально эффективно: самое маленькое подходящее печенье отдавать самому «дешевому» (неприхотливому) ребенку.

```python
from typing import List

class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        # 1. Сортируем детей по жадности, а печенье по размеру
        g.sort()
        s.sort()
        
        child_i = 0  # Указатель на текущего самого голодного ребенка
        cookie_j = 0  # Указатель на текущее самое маленькое печенье
        
        # 2. Жадный проход
        while child_i < len(g) and cookie_j < len(s):
            # Если текущее печенье достаточно для текущего ребенка
            if s[cookie_j] >= g[child_i]:
                # Жадный выбор: отдаем это печенье и переходим к следующему ребенку
                child_i += 1
            
            # В любом случае переходим к следующему печенью (так как текущее 
            # либо уже съедено, либо слишком мало для этого ребенка и всех более жадных)
            cookie_j += 1
            
        return child_i
```

**Пояснение**: 
Почему это жадный алгоритм? Потому что мы принимаем локально оптимальное решение: отдаем ребенку минимально возможное печенье, которое его удовлетворит. Мы не «тратим» большое печенье на ребенка с маленьким запросом, приберегая его для более «жадных» детей.
