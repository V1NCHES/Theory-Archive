# Куча (Heap) и Очередь с приоритетом (Priority Queue)

**Куча (Heap)** — это специализированная древовидная структура данных, которая удовлетворяет свойству кучи.

## 1. Основные понятия

### Типы куч:
1.  **Min-Heap (Минимальная куча):** Значение в любом узле меньше или равно значениям его потомков. Корень содержит **минимальный** элемент.
2.  **Max-Heap (Максимальная куча):** Значение в любом узле больше или равно значениям его потомков. Корень содержит **максимальный** элемент.

### Свойства:
- **Завершенность (Complete Binary Tree):** Куча всегда является полным бинарным деревом — все уровни заполнены, кроме, возможно, последнего, который заполняется слева направо.
- **Эффективность:** Позволяет быстро находить экстремум (минимум или максимум).

---

## 2. Представление в памяти

Хотя куча — это дерево, её обычно реализуют с помощью **массива** для экономии памяти.
Для узла с индексом `i`:
- **Левый потомок:** `2 * i + 1`
- **Правый потомок:** `2 * i + 2`
- **Родитель:** `(i - 1) // 2`

---

## 3. Основные операции и их сложность

| Операция | Описание | Сложность |
| :--- | :--- | :--- |
| **Peek** | Получить корень (min или max) без удаления | $O(1)$ |
| **Insert** | Добавить новый элемент (добавление в конец + Sift Up) | $O(\log n)$ |
| **Extract Min/Max** | Удалить корень (замена последним элементом + Sift Down) | $O(\log n)$ |
| **Heapify** | Преобразование произвольного массива в кучу | $O(n)$ |

---

## 4. Очередь с приоритетом (Priority Queue)

**Очередь с приоритетом** — это абстрактный тип данных, где каждый элемент имеет «приоритет». Элемент с наивысшим приоритетом обслуживается первым. 
Куча является наиболее эффективным способом реализации очереди с приоритетом.

---

## 5. Реализация в Python (`heapq`)

В Python модуль `heapq` реализует **Min-Heap**.

```python
import heapq

# Создание кучи из списка
nums = [5, 7, 9, 1, 3]
heapq.heapify(nums) 
print(f"Куча: {nums}")

# Добавление элемента
heapq.heappush(nums, 4)

# Извлечение минимального
min_val = heapq.heappop(nums)
print(f"Извлеченный минимум: {min_val}")
print(f"Куча после удаления: {nums}")

# n самых больших/маленьких элементов
largest = heapq.nlargest(3, nums)
smallest = heapq.nsmallest(2, nums)
print(f"3 самых больших: {largest}")
```

### Как сделать Max-Heap в Python?
Поскольку `heapq` поддерживает только Min-Heap, для Max-Heap значения обычно умножают на `-1`.

```python
import heapq
max_heap = [5, 7, 9, 1, 3]
max_heap = [-x for x in max_heap]
heapq.heapify(max_heap)

# Чтобы получить реальное значение:
actual_max = -heapq.heappop(max_heap)
print(f"Максимальный элемент: {actual_max}")
```

## 6. Когда использовать (LeetCode Patterns)

1.  **Top K Elements:** Найти K самых больших/маленьких элементов (сложность $O(n \log k)$).
2.  **K-way Merge:** Слияние нескольких отсортированных списков.
3.  **Running Median:** Поддержание медианы в потоке данных (используются две кучи: Min и Max).
4.  **Алгоритм Дейкстры:** Поиск кратчайшего пути в графе.

Вам дан целочисленный массив scoreразмером n, где score[i]— это результат спортсмена в соревновании. Гарантируется уникальность всех результатов .i^{th}

Распределение спортсменов по местам происходит на основе набранных ими баллов: спортсмен, занявший первое место, набрал наибольшее количество баллов, спортсмен, занявший второе место, набрал наибольшее количество баллов, и так далее. Место каждого спортсмена определяет его рейтинг:1st2nd2nd

Место , которое занимает спортсмен в рейтинге, — это .1st"Gold Medal"
Место , которое занимает спортсмен в рейтинге, — это .2nd"Silver Medal"
Место , которое занимает спортсмен в рейтинге, — это .3rd"Bronze Medal"
Для спортсмена, занявшего призовое место, его место определяется номером занятого места (например, место спортсмена, занявшего призовое место, обозначается как ).4thnthxth"x"
Возвращает массив answer размером , nгде answer[i]— ранг спортсмена .ith

```python
from typing import List
import heapq
class Solution:
    def findRelativeRanks(self, score: List[int]) -> List[str]:
        n = len(score)
        heap = []
    # 1. Кладем в кучу (минус балл, индекс)
        for i in range(n):
            heapq.heappush(heap, (-score[i], i))
    
        ans = [0] * n
        rank_names = ["Gold Medal", "Silver Medal", "Bronze Medal"]
    
    # 2. Достаем по одному (сначала самые большие баллы)
        for i in range(1, n + 1):
             _, idx = heapq.heappop(heap)
             if i <= 3:
                  ans[idx] = rank_names[i-1]
             else:
                 ans[idx] = str(i)
        return ans




sol = Solution()
print(f"result: {sol.findRelativeRanks([5,4,3,2,1])}")
```

703. Крупнейший элемент в потоке. https://leetcode.com/problems/kth-largest-element-in-a-stream/description/

Вы работаете в приемной комиссии университета и должны в режиме реального времени отслеживать kthнаивысшие баллы, полученные абитуриентами на вступительных экзаменах. Это помогает динамически определять проходные баллы для собеседований и зачисления по мере того, как новые абитуриенты предоставляют свои результаты.

Вам поручено реализовать класс, который для заданного целого числа  kподдерживает поток результатов тестов и непрерывно возвращает -й  kнаивысший результат теста  после  отправки нового результата. В частности, мы ищем k-й наивысший результат в отсортированном списке всех результатов.

Реализуйте  KthLargestкласс:

KthLargest(int k, int[] nums)Инициализирует объект целочисленным значением kи потоком результатов тестов  nums.
int add(int val)Добавляет новый результат теста  valв поток и возвращает элемент, представляющий наибольший элемент в пуле результатов тестов на данный момент.kth

```python
from typing import List
import heapq

class KthLargest:
    def __init__(self, k: int, nums: List[int]):
        self.k = k
        self.heap = nums
        heapq.heapify(self.heap)
        # Оставляем в куче только k самых больших элементов
        while len(self.heap) > k:
            heapq.heappop(self.heap)

    def add(self, val: int) -> int:
        heapq.heappush(self.heap, val)
        # Если элементов стало больше k, удаляем самый маленький
        if len(self.heap) > self.k:
            heapq.heappop(self.heap)
        # Корень Min-Heap — это k-й по величине элемент
        return self.heap[0]

# Тестирование:
k = 3
nums = [4, 5, 8, 2]
obj = KthLargest(k, nums)
print(f"add(3): {obj.add(3)}")   # returns 4
print(f"add(5): {obj.add(5)}")   # returns 5
print(f"add(10): {obj.add(10)}") # returns 5
print(f"add(9): {obj.add(9)}")   # returns 8
print(f"add(4): {obj.add(4)}")   # returns 8
```
