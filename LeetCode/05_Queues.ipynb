# Очередь (Queue)

**Очередь** — это линейная структура данных, работающая по принципу **FIFO (First-In, First-Out)**: «первым пришел — первым ушел».

## 1. Основные понятия

Представьте очередь в магазине: тот, кто встал в неё первым, первым будет обслужен и первым её покинет.

### Основные операции:
1.  **Enqueue (Push):** Добавление элемента в конец (хвост) очереди.
2.  **Dequeue (Pop):** Удаление элемента из начала (головы) очереди.
3.  **Peek (Front):** Получение значения первого элемента без его удаления.
4.  **isEmpty:** Проверка очереди на пустоту.

### Сложность операций:
| Операция | Сложность |
| :--- | :--- |
| Добавление (Enqueue) | $O(1)$ |
| Удаление (Dequeue) | $O(1)$ |
| Доступ к первому (Peek) | $O(1)$ |

---

## 2. Реализация в Python

### Плохой способ: `list`
Использование обычного списка `list` для очереди неэффективно. 
- `list.append()` — это $O(1)$.
- `list.pop(0)` — это **$O(n)$**, так как все остальные элементы должны сдвинуться на одну позицию влево.

### Правильный способ: `collections.deque` (Double-Ended Queue)
В Python для очередей идеально подходит `deque`. Она реализована как двусвязный список, что позволяет добавлять и удалять элементы с обоих концов за $O(1)$.

```python
from collections import deque

# Создание очереди
queue = deque()

# 1. Добавление в конец (Enqueue)
queue.append("первый")
queue.append("второй")
queue.append("третий")

print(f"Очередь: {queue}")

# 2. Удаление из начала (Dequeue)
first = queue.popleft()
print(f"Удален: {first}")
print(f"Осталось: {queue}")

# 3. Посмотреть первый элемент (Peek)
if queue:
    print(f"Первый в очереди: {queue[0]}")

# 4. Проверка размера
print(f"Размер: {len(queue)}")
```

## 3. Типы очередей

1.  **Простая очередь:** Стандартный FIFO.
2.  **Циклическая очередь:** Последняя позиция соединена с первой (используется в буферах фиксированного размера).
3.  **Двусторонняя очередь (Deque):** Позволяет добавлять/удалять элементы с обоих концов.
4.  **Очередь с приоритетом (Priority Queue):** Элементы извлекаются на основе приоритета, а не порядка добавления (обычно реализуется через Кучу/Heap).

## 4. Паттерны применения (LeetCode)

### 1. BFS (Breadth-First Search) — Обход в ширину
Очередь — это ключевая структура для BFS в графах и деревьях. Мы посещаем всех соседей текущего узла, прежде чем переходить к следующему уровню.

### 2. Скользящее окно (Sliding Window)
Часто используется для задач, где нужно отслеживать элементы в определенном интервале.

### 3. Управление задачами (Task Scheduling)
Очередь используется для обработки запросов в порядке поступления (например, очередь печати или обработка HTTP-запросов).

Реализуйте MyStackкласс:

void push(int x)Помещает элемент x на вершину стека.
int pop()Удаляет элемент, находящийся на вершине стека, и возвращает его.
int top()Возвращает элемент, находящийся на вершине стека.
boolean empty()Возвращает значение, trueесли стек пуст, falseв противном случае — в противном случае.

```python
from collections import deque


class MyStack:

    def __init__(self):
        # Создание очереди
        self.queue = deque()

    def push(self, x: int) -> None:
        self.queue.append(x)   

    def pop(self) -> int:
        return self.queue.popleft()

    def top(self) -> int:
        print()

    def empty(self) -> bool:
        print()
```
