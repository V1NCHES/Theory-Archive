# Жадные алгоритмы (Greedy Algorithms)

**Жадный алгоритм** — это метод решения задач, основанный на принятии локально оптимального решения на каждом шаге в надежде, что это приведет к глобально оптимальному решению.

### Основные принципы:
1.  **Жадный выбор**: На каждом шаге мы выбираем лучший вариант «здесь и сейчас», не задумываясь о будущих последствиях.
2.  **Оптимальная подструктура**: Оптимальное решение всей задачи строится из оптимальных решений её подзадач.

### Когда жадность работает?
Жадные алгоритмы работают не всегда. Для их применения нужно доказать две вещи:
*   **Принцип жадного выбора**: Локально лучший выбор не закроет путь к глобальному оптимуму.
*   **Оптимальность подзадач**: Решив подзадачу жадно, мы упростим исходную задачу до такой же, но меньшего размера.

### Типичные стратегии:
*   **Сортировка**: Часто жадное решение начинается с сортировки данных по какому-то признаку (цене, весу, времени).
*   **Подсчет частот**: Выбор наиболее или наименее часто встречающихся элементов.

### 409. Самый длинный палиндром https://leetcode.com/problems/longest-palindrome/

Если задана строка `s`, состоящая из строчных или заглавных букв, верните длину самого длинного палиндрома что можно построить из этих букв..

Буквы чувствительны к регистру , например, "Aa"не считается палиндромом.

**Теория для задачи**: Палиндром строится из пар одинаковых букв. Если буква встречается 5 раз, мы можем жадно взять 4 из них (2 пары) для палиндрома. Если останутся лишние буквы, одну из них можно поставить в центр.


Метод/операция	Описание	Пример
elements()	Возвращает итератор, повторяющий элемент столько раз, сколько его счётчик	list(c.elements())
most_common(n)	Возвращает n самых частых элементов	c.most_common(2) → [('a', 3), ('b', 2)]
total() (Python 3.10+)	Сумма всех счётчиков	c.total() → 6
update(iterable/mapping)	Добавляет счётчики (не заменяет — складывает)	c.update("abc")
subtract(iterable/mapping)	Вычитает счётчики	c.subtract("abc")
items(), keys(), values()	Как у обычного словаря	—

```python
from collections import Counter

class Solution:
    def longestPalindrome(self, s: str) -> int:
        # 1. Считаем сколько раз встречается каждая буква
        counts = Counter(s)
        length = 0
        has_odd = False
        
        # 2. Жадный выбор: забираем все пары из каждой буквы
        for count in counts.values():
            # Если букв четное количество, берем все
            # Если нечетное (например 3), берем (3 - 1) = 2
            length += (count // 2) * 2
            
            # Если осталась хотя бы одна лишняя буква, помечаем это
            if count % 2 == 1:
                has_odd = True
        
        # 3. Если есть хоть одна лишняя буква, ставим её в центр палиндрома (+1 к длине)
        return length + 1 if has_odd else length
```

**Пояснение**: 
Мы действуем жадно, так как на каждом шаге (для каждой буквы) мы максимизируем её вклад в длину палиндрома, забирая все возможные пары. Нам не нужно знать, какие буквы мы возьмем дальше — пары всегда выгодны.

### 455. Назначение файлов cookie https://leetcode.com/problems/assign-cookies/

У каждого ребенка есть коэффициент жадности `g[i]`, а у каждого печенья — размер `s[j]`. Цель — максимизировать количество довольных детей.

Если s[j] >= g[i], мы можем назначить печенье jдочернему элементу i, и дочерний элемент iбудет доволен. Ваша цель — максимизировать количество довольных дочерних элементов и вывести максимальное их количество.

**Теория для задачи**: Это классическая жадная задача. Чтобы удовлетворить максимум детей, нужно тратить ресурсы (печенье) максимально эффективно: самое маленькое подходящее печенье отдавать самому «дешевому» (неприхотливому) ребенку.

```python
from typing import List

class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        # 1. Сортируем детей по жадности, а печенье по размеру
        g.sort()
        s.sort()
        
        child_i = 0  # Указатель на текущего самого голодного ребенка
        cookie_j = 0  # Указатель на текущее самое маленькое печенье
        
        # 2. Жадный проход
        while child_i < len(g) and cookie_j < len(s):
            # Если текущее печенье достаточно для текущего ребенка
            if s[cookie_j] >= g[child_i]:
                # Жадный выбор: отдаем это печенье и переходим к следующему ребенку
                child_i += 1
            
            # В любом случае переходим к следующему печенью (так как текущее 
            # либо уже съедено, либо слишком мало для этого ребенка и всех более жадных)
            cookie_j += 1
            
        return child_i
```

**Пояснение**: 
Почему это жадный алгоритм? Потому что мы принимаем локально оптимальное решение: отдаем ребенку минимально возможное печенье, которое его удовлетворит. Мы не «тратим» большое печенье на ребенка с маленьким запросом, приберегая его для более «жадных» детей.

561. Array Partition https://leetcode.com/problems/array-partition/description/

Дан массив целых numsчисел 2n. Сгруппируйте эти числа в nпары таким образом, чтобы сумма всех чисел была максимальной . Верните максимальную сумму .(a1, b1), (a2, b2), ..., (an, bn) min(ai, bi)i

```python
from typing import List
class Solution:
    def arrayPairSum(self, nums: List[int]) -> int:
        nums.sort()
        sum = 0
        #print(nums)
        for iter in nums[::2]:
            #print(iter)
            sum+=iter
        return sum
            

sol = Solution()
print(sol.arrayPairSum([1,4,3,2]))
```

605. Можно поставить цветы. https://leetcode.com/problems/can-place-flowers/

У вас есть длинная клумба, на которой часть участков засажена цветами, а часть — нет. Однако цветы нельзя сажать на соседних участках.

Дан массив целых чисел, flowerbedсодержащий 0's' и 1's', где 's' 0означает пустой массив, а 1's' означает непустой массив, и целое число 's '. nВерните значение true , если n в массиве можно посадить новые цветы, flowerbed не нарушая правило отсутствия соседних цветов, и false значение в противном случае .

```python
from typing import List
class Solution:
    def canPlaceFlowers(self, flowerbed: List[int], n: int) -> bool:
        l = len(flowerbed)
        if(n==0):
            return True
        if(l==1 and flowerbed[0] == 0 and n == 1):
            return True
        elif(l==1 and flowerbed[0] == 0):
            return False

        if(l == 2 and flowerbed[0] == 0 and flowerbed[1] == 0 and n == 1):
            return True
        elif(l == 2 and flowerbed[0] == 0 and flowerbed[1] == 0):
            return False
        
        if(flowerbed[0] == 0 and flowerbed[1] == 0):
            flowerbed[0] = 1
            n-=1
            if(n==0):
                    return True

        for i in range(1, l - 1, 1):
            #print(i, flowerbed)
            if(flowerbed[i-1] == 0 and flowerbed[i] == 0 and flowerbed[i+1] == 0):
                flowerbed[i] = 1
                i+=1
                n-=1
                if(n==0):
                    return True

        if(flowerbed[-1] == 0 and flowerbed[-2] == 0):
            flowerbed[-1] = 1
            n-=1
            if(n==0):
                    return True
        
        return False

sol = Solution()
print(sol.canPlaceFlowers([1,0,0,0,1,0,0], 2))
```

680. Действительный палиндром II https://leetcode.com/problems/valid-palindrome-ii/description/

Если задана строка s, верните значение, true если она sможет быть палиндромом после удаления не более одного символа .

```python
class Solution:
    def validPalindrome(self, s: str) -> bool:
        def is_palindrome_range(left: int, right: int) -> bool:
            """Проверяет, является ли подстрока s[left:right+1] палиндромом"""
            while left < right:
                if s[left] != s[right]:
                    return False
                left += 1
                right -= 1
            return True
        
        left, right = 0, len(s) - 1
        
        while left < right:
            if s[left] != s[right]:
                # Пробуем удалить левый символ ИЛИ правый символ
                return (is_palindrome_range(left + 1, right) or 
                        is_palindrome_range(left, right - 1))
            left += 1
            right -= 1
        
        return True

sol = Solution()
print(sol.validPalindrome("eceec"))
```

860. Лимонадная сдача https://leetcode.com/problems/lemonade-change/description/

В киоске по продаже лимонада каждый лимонад стоит [цена] $5. Покупатели стоят в очереди, чтобы купить у вас лимонад, и заказывают по одному лимонаду за раз (в порядке, указанном на чеках). Каждый покупатель покупает только один лимонад и расплачивается купюрой [цена] $5, [цена $10] или $20[цена]. Вы должны выдать каждому покупателю правильную сдачу, чтобы в итоге покупатель заплатил [цена] $5.

Обратите внимание, что поначалу у вас нет сдачи.

Дана целочисленная матрица, billsгде bills[i]— сумма счета , который оплачивает клиент. Верните значение, если вы можете выдать каждому клиенту правильную сдачу, или значение, если нет .ithtrue false

```python
from typing import List
class Solution:
    def lemonadeChange(self, bills: List[int]) -> bool:
        f = t = 0
        for i in bills:
            if i == 5:
                f +=1
            elif i == 10:
                if f == 0:
                    return False
                t += 1
                f -= 1
            else:
                if t > 0 and f >0:
                    t -= 1
                    f -= 1
                elif f >=3:
                    f -= 3
                else:
                    return False
        return True

sol = Solution()
print(sol.lemonadeChange([5,5,5,10,20]))
```
