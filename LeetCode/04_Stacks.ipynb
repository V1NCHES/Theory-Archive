# Стек (Stack)

**Стек** — это линейная структура данных, которая следует принципу **LIFO** (Last In, First Out — «последним пришёл, первым ушёл»).

Это можно представить как стопку тарелок: вы кладёте новую тарелку сверху и снимаете тоже только верхнюю тарелку.

### Основные операции:
1.  **Push**: Добавление элемента в вершину стека — `O(1)`.
2.  **Pop**: Удаление и возврат элемента из вершины стека — `O(1)`.
3.  **Peek (Top)**: Возврат элемента из вершины без его удаления — `O(1)`.
4.  **isEmpty**: Проверка, пуст ли стек — `O(1)`.
5.  **Size**: Возврат количества элементов в стеке — `O(1)`.

### Реализация в Python:
1.  **List (список)**: Можно использовать `append()` и `pop()`. Это самый простой способ.
2.  **collections.deque**: Двусторонняя очередь. В отличие от списка, она обеспечивает `O(1)` для добавления и удаления элементов с обоих концов и более эффективна по памяти в некоторых случаях.

```python
from collections import deque

stack = deque()

stack.append(1)  # Push
stack.append(2)
stack.append(3)

print(f"Вершина стека: {stack[-1]}")  # Peek: 3
print(f"Удален элемент: {stack.pop()}")  # Pop: 3
print(f"Стек пуст? {not stack}") # isEmpty: False
```

### 94. Binary Tree Inorder Traversal https://leetcode.com/problems/binary-tree-inorder-traversal/

**Условие**: Имея представление rootо бинарном дереве, верните обход его узлов в порядке возрастания значений .

**Использование стека**: Мы можем имитировать рекурсию с помощью собственного стека. Это позволяет избежать переполнения стека вызовов и наглядно показывает, как работает DFS.

```python
from typing import List, Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        res, stack = [], []
        curr = root
        
        # Итеративный подход с использованием стека (LNR)
        while curr or stack:
            # 1. Спускаемся до самого левого узла
            while curr:
                stack.append(curr)
                curr = curr.left
            
            # 2. Обрабатываем узел (корень поддерева)
            curr = stack.pop()
            res.append(curr.val)
            
            # 3. Переходим к правому поддереву
            curr = curr.right
            
        return res

# Пример для теста: [1, null, 2, 3]
root = TreeNode(1)
root.right = TreeNode(2)
root.right.left = TreeNode(3)

sol = Solution()
print(f"Inorder Traversal result: {sol.inorderTraversal(root)}")
```
