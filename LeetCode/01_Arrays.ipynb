# Массивы (Arrays)

**Массив** — это структура данных, представляющая собой упорядоченный набор элементов, доступ к которым осуществляется по индексу. В памяти элементы массива обычно располагаются в виде непрерывного блока.

### Основные характеристики:
*   **Доступ по индексу**: $O(1)$ — мы можем мгновенно получить любой элемент, зная его индекс.
*   **Поиск**: $O(n)$ в общем случае (линейный поиск). Если массив отсортирован, можно использовать бинарный поиск с $O(\log n)$.
*   **Вставка/Удаление**: $O(n)$ — так как при добавлении или удалении элемента в середину или начало массива, остальные элементы необходимо сдвинуть.
*   **Вставка в конец**: $O(1)$ в среднем для динамических массивов.

### Динамические массивы (Python `list`):
В Python `list` реализован как динамический массив. Это означает, что он может изменять свой размер автоматически при добавлении элементов.
*   При заполнении выделенной памяти Python выделяет новый, более крупный блок памяти и копирует туда старые элементы.
*   Это позволяет амортизировать стоимость вставки в конец до $O(1)$.

### Основные приемы решения задач с массивами:
1.  **Два указателя (Two Pointers)**: движение навстречу друг другу или в одном направлении.
2.  **Скользящее окно (Sliding Window)**: для задач на подмассивы.
3.  **Префиксные суммы (Prefix Sums)**: для быстрого вычисления суммы на отрезках.
4.  **Сортировка**: часто упрощает решение задачи.
5.  **Хеш-таблицы**: для быстрого поиска дополнений (как в задаче Two Sum).

### 1. Две суммы https://leetcode.com/problems/two-sum/

**Теория**: В массивах часто приходится искать пары элементов. Использование хеш-таблицы позволяет избежать вложенного цикла и снизить сложность с $O(n^2)$ до $O(n)$.

Дан массив целых чисел nums и целое число target. Верните индексы этих двух чисел, сумма которых равнаtarget .

Можно предположить, что для каждого входного значения существует ровно одно решение , и нельзя использовать один и тот же элемент дважды.

Вы можете вернуть ответ в любом порядке.

```python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
        return []


solution = Solution()
print(solution.twoSum([1,2,3,4,5], 3))
```

**Пояснение**: Мы проходим по списку один раз, на каждом шаге сохраняя число и его индекс. Поиск в словаре работает за $O(1)$.

### 26. Удалите дубликаты из отсортированного массива https://leetcode.com/problems/remove-duplicates-from-sorted-array/

**Теория**: Метод **двух указателей**. Один указатель (`k`) идет по уникальным позициям, второй (`i`) сканирует весь массив. Это позволяет изменять массив «на месте» без выделения новой памяти.

```python
from typing import List
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        if not nums: return 0
        k = 1  # Позиция для следующего уникального элемента
        for i in range(1, len(nums)):
            # Если нашли новый элемент (не равный предыдущему)
            if nums[i] != nums[i - 1]:
                nums[k] = nums[i]  # Переносим его вперед
                k += 1
        return k
```

**Пояснение**: Благодаря тому, что массив отсортирован, все дубликаты идут подряд. Мы просто «перепрыгиваем» через них, копируя только уникальные значения в начало массива.

### 35. Поиск позиции вставки https://leetcode.com/problems/search-insert-position/

**Теория**: **Бинарный поиск**. В отсортированных массивах для поиска всегда следует использовать бинарный поиск, чтобы получить сложность $O(\log n)$.

```python
from typing import List
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1
        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target: return mid
            if nums[mid] < target: left = mid + 1
            else: right = mid - 1
        return left
```

**Пояснение**: Мы постоянно делим область поиска пополам. Если число не найдено, `left` окажется именно в той позиции, куда число следовало бы вставить.

### 88. Объединение отсортированного массива https://leetcode.com/problems/merge-sorted-array/

**Теория**: **Два указателя с конца**. Чтобы не сдвигать элементы (что долго), мы начинаем заполнять `nums1` с самых больших элементов, двигаясь от конца к началу.

Вам даны два массива целых чисел nums1  и nums2, отсортированные в порядке убывания , и два целых числа m и n, представляющие количество элементов в nums1 и nums2 соответственно.

Объединить nums1 и nums2в один массив, отсортированный в порядке убывания .

Итоговый отсортированный массив не должен возвращаться функцией, а должен храниться внутри массиваnums1 . Для этого массив nums1имеет длину m + n, где первые m элементы обозначают элементы, которые должны быть объединены, а последние nэлементы устанавливаются в значение 0и должны игнорироваться. nums2Массив имеет длину n.

```python
from typing import List

class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        p1, p2, p = m - 1, n - 1, m + n - 1
        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                nums1[p] = nums1[p1]
                p1 -= 1
            else:
                nums1[p] = nums2[p2]
                p2 -= 1
            p -= 1
        # Докопируем остатки из nums2, если они есть
        if p2 >= 0:
            nums1[:p2 + 1] = nums2[:p2 + 1]

solution = Solution()
print(solution.merge([1,2,3,0,0,0], 3, [2,5,6], 3))
```

**Пояснение**: Заполнение с конца позволяет использовать пустое место в `nums1` без риска перезаписать еще не обработанные данные.

### 136. Единый номер https://leetcode.com/problems/single-number/

**Теория**: **Битовые манипуляции (XOR)**. Исключающее ИЛИ обладает свойством: $A \oplus A = 0$ и $A \oplus 0 = A$. Это позволяет найти уникальный элемент без использования памяти.

```python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        res = 0
        for n in nums: res ^= n
        return res
```

**Пояснение**: Все парные числа «уничтожат» друг друга через XOR, останется только то число, у которого нет пары.
