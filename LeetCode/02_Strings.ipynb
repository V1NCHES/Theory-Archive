# Строки (Strings)

**Строка** — это неизменяемая (в Python) последовательность символов. В памяти строки хранятся как массивы символов.

### Основные характеристики:
*   **Индексация**: $O(1)$ — доступ к символу по его позиции.
*   **Длина**: $O(1)$ в Python (хранится в метаданных объекта).
*   **Конкатенация**: $O(n + m)$ — так как строки неизменяемы, при сложении создается новая строка, куда копируются данные обеих исходных.
*   **Поиск подстроки**: $O(n)$ в среднем (реализация Python оптимизирована).

### Особенности в Python:
*   **Неизменяемость (Immutability)**: Любая модификация строки (замена символа, удаление) создает новый объект в памяти.
*   **Эффективное объединение**: Вместо использования `+` в цикле, лучше собирать части в список и использовать `''.join(list)`.
*   **Методы**: `.split()`, `.join()`, `.strip()`, `.replace()`, `.find()`, `.lower()`, `.upper()`.

### Основные приемы решения задач:
1.  **Два указателя**: для задач на палиндромы, разворот строки или поиск.
2.  **Скользящее окно**: для поиска подстрок с определенными свойствами.
3.  **Хеш-таблицы**: для подсчета частоты символов (задачи на анаграммы).
4.  **Стек**: для задач на вложенные структуры (например, скобки).

### 13. Римский в целое число https://leetcode.com/problems/roman-to-integer/

Римские цифры обозначаются семью различными символами: `I, V, X, L, C, D` и `M`.

| Символ | Значение |
| :--- | :--- |
| I | 1 |
| V | 5 |
| X | 10 |
| L | 50 |
| C | 100 |
| D | 500 |
| M | 1000 |

Дана римская цифра, преобразуйте её в целое число.

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        m = {'I':1, 'V':5, 'X':10, 'L':50, 'C':100, 'D':500, 'M':1000}
        total = 0
        for i in range(len(s)):
            if i < len(s) - 1 and m[s[i]] < m[s[i+1]]:
                total -= m[s[i]]
            else:
                total += m[s[i]]
        return total

# Пример использования:
sol = Solution()
print(sol.romanToInt("LVIII"))  # Ожидаемый результат: 58
```

### 14. Самый длинный общий префикс https://leetcode.com/problems/longest-common-prefix/

Напишите функцию для поиска самой длинной общей префиксной строки среди массива строк.

Если общего префикса нет, вернуть пустую строку `""`.

```python
from typing import List

class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        if not strs:
            return ""
        
        prefix = strs[0]
        for s in strs[1:]:
            while not s.startswith(prefix):
                prefix = prefix[:-1]
                if not prefix: return ""
        return prefix

# Пример использования:
sol = Solution()
print(sol.longestCommonPrefix(["flower","flow","flight"]))  # Ожидаемый результат: "fl"
```

### 20. Допустимые скобки https://leetcode.com/problems/valid-parentheses/

Дана строка `s`, содержащая только символы `(`, `)`, `{`, `}`, `[` и `]`. Определите, является ли входная строка допустимой.

Строка считается допустимой, если:
1. Открытые скобки должны быть закрыты скобками того же типа.
2. Открытые скобки должны быть закрыты в правильном порядке.

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        mapping = {")": "(", "}": "{", "]": "["}
        
        for char in s:
            if char in mapping:
                top_element = stack.pop() if stack else '#'
                if mapping[char] != top_element:
                    return False
            else:
                stack.append(char)
        
        return not stack

# Пример использования:
sol = Solution()
print(sol.isValid("()[]{}"))  # Ожидаемый результат: True
```

### 28. Индекс первого вхождения в строке https://leetcode.com/problems/find-the-index-of-the-first-occurrence-in-a-string/

Даны две строки: `needle` (игла) и `haystack` (стог сена). Верните индекс первого вхождения строки `needle` в строку `haystack`, или `-1`, если `needle` не является частью `haystack`.

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        if not needle:
            return 0
        return haystack.find(needle)

# Пример использования:
sol = Solution()
print(sol.strStr("sadbutsad", "sad"))  # Ожидаемый результат: 0
```

### 58. Длина последнего слова https://leetcode.com/problems/length-of-last-word/

Дана строка `s`, состоящая из слов и пробелов. Верните длину последнего слова в этой строке.
Слово — это максимальная подстрока, состоящая только из символов, не являющихся пробелами.

```python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        words = s.strip().split()
        if not words:
            return 0
        return len(words[-1])

# Пример использования:
sol = Solution()
print(sol.lengthOfLastWord("   fly me   to   the moon  "))  # Ожидаемый результат: 4
```

### 67. Сложение двоичных чисел https://leetcode.com/problems/add-binary/

Даны две двоичные строки `a` и `b`. Верните их сумму в виде двоичной строки.

```python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        return bin(int(a, 2) + int(b, 2))[2:]

# Пример использования:
sol = Solution()
print(sol.addBinary("1010", "1011"))  # Ожидаемый результат: "10101"
```

### 125. Действительный палиндром https://leetcode.com/problems/valid-palindrome/

Фраза называется палиндромом, если после преобразования всех заглавных букв в строчные и удаления всех небуквенно-цифровых символов она читается одинаково как в прямом, так и в обратном порядке.

Если задана строка `s`, вернуть `true`, если она является палиндромом, или `false` в противном случае.

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        filtered_chars = [char.lower() for char in s if char.isalnum()]
        return filtered_chars == filtered_chars[::-1]

# Пример использования:
sol = Solution()
print(sol.isPalindrome("A man, a plan, a canal: Panama"))  # Ожидаемый результат: True
```

### 168. Заголовок столбца таблицы Excel https://leetcode.com/problems/excel-sheet-column-title/

Дано целое число `columnNumber`, верните соответствующий заголовок столбца в том виде, в котором он отображается в Excel (например, 1 -> A, 2 -> B, 26 -> Z, 27 -> AA).

```python
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        result = []
        while columnNumber > 0:
            columnNumber -= 1  # Смещение для 0-индексации
            result.append(chr(columnNumber % 26 + ord('A')))
            columnNumber //= 26
        return "".join(result[::-1])

# Пример использования:
sol = Solution()
print(sol.convertToTitle(28))  # Ожидаемый результат: "AB"
```
