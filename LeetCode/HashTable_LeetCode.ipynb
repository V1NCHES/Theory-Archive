# Хеш-таблицы (Hash Tables)

**Хеш-таблица** — это структура данных, реализующая интерфейс ассоциативного массива (пары "ключ-значение"). В Python хеш-таблицы реализованы в виде словарей (`dict`) и множеств (`set`).

### Основные характеристики:
*   **Доступ по ключу**: $O(1)$ в среднем.
*   **Вставка/Удаление**: $O(1)$ в среднем.
*   **Поиск**: $O(1)$ в среднем.
*   **Худший случай**: $O(n)$ при большом количестве коллизий.

### Как это работает:
1.  **Хеш-функция**: Превращает ключ в индекс массива.
2.  **Коллизии**: Ситуации, когда два разных ключа имеют один и тот же хеш.
3.  **Ключи**: Должны быть хэшируемыми (неизменяемыми: числа, строки, кортежи).

### Когда использовать хеш-таблицы:
1.  **Быстрая проверка наличия**: Проверить, встречался ли элемент ранее.
2.  **Подсчет частоты**: Посчитать, сколько раз встречается каждый символ или число.
3.  **Поиск дополнений**: Например, поиск `target - x` в задаче Two Sum.
4.  **Устранение дубликатов**: Использование `set` для хранения уникальных значений.

### 1. Две суммы https://leetcode.com/problems/two-sum/

Дан массив целых чисел `nums` и целое число `target`. Верните индексы этих двух чисел, сумма которых равна `target`.

**Теория для задачи**: Вместо того чтобы проверять все пары чисел ($O(n^2)$), мы можем использовать хеш-таблицу для хранения уже просмотренных чисел и их индексов. Это позволяет найти «дополнение» (`target - current_number`) за $O(1)$.

```python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # seen будет хранить {число: его_индекс}
        seen = {}
        for i, num in enumerate(nums):
            # Вычисляем число, которое нам нужно найти
            complement = target - num
            
            # Проверяем, видели ли мы это число раньше
            if complement in seen:
                # Если да, возвращаем индекс старого числа и текущий индекс
                return [seen[complement], i]
            
            # Если нет, запоминаем текущее число и его индекс
            seen[num] = i
        return []
```

**Пояснение**: 
Алгоритм проходит по массиву один раз. На каждом шаге мы спрашиваем хеш-таблицу: «Есть ли у тебя число `X`?». Благодаря свойствам хеш-таблицы, ответ мы получаем мгновенно. Это сокращает сложность с квадратичной до линейной.

### 13. Римский в целое число https://leetcode.com/problems/roman-to-integer/

**Теория для задачи**: Хеш-таблица здесь выступает в роли «словаря соответствий». Вместо длинных цепочек `if-elif`, мы создаем константную таблицу ценности каждого символа.

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        # Хеш-таблица (словарь) для быстрого доступа к значениям
        values = {
            'I': 1, 'V': 5, 'X': 10, 'L': 50, 
            'C': 100, 'D': 500, 'M': 1000
        }
        total = 0
        i = 0
        while i < len(s):
            # Если текущий символ меньше следующего (например, IV), вычитаем его
            if i + 1 < len(s) and values[s[i]] < values[s[i+1]]:
                total += values[s[i+1]] - values[s[i]]
                i += 2  # Пропускаем оба символа
            else:
                total += values[s[i]]
                i += 1
        return total
```

**Пояснение**: 
Использование словаря делает код чистым и легко расширяемым. Если бы добавились новые символы, нам нужно было бы просто дописать их в таблицу `values`.

### 141. Цикл связанных списков https://leetcode.com/problems/linked-list-cycle/

**Теория для задачи**: Мы используем свойство хеш-таблицы хранить ссылки на объекты. В Python `set()` сохраняет уникальные хеши объектов (в данном случае — адреса узлов в памяти).

```python
class Solution:
    def hasCycle(self, head) -> bool:
        # Множество для хранения уже посещенных узлов
        visited = set()
        current = head
        while current:
            # Если узел уже есть в visited, значит мы вернулись по кругу
            if current in visited:
                return True
            # Запоминаем текущий узел
            visited.add(current)
            current = current.next
        return False
```

**Пояснение**: 
Этот подход требует $O(n)$ дополнительной памяти (для хранения узлов), но он максимально прост в понимании: «если я тут уже был — значит, я хожу по кругу».

### 160. Пересечение двух связанных списков https://leetcode.com/problems/intersection-of-two-linked-lists/

**Теория для задачи**: Аналогично задаче на цикл, мы используем хеш-таблицу как «память». Сохранив весь первый список, мы можем мгновенно проверить каждый узел второго списка на предмет вхождения в первый.

```python
class Solution:
    def getIntersectionNode(self, headA, headB):
        # 1. Запоминаем все узлы первого списка
        nodes_in_a = set()
        curr = headA
        while curr:
            nodes_in_a.add(curr)
            curr = curr.next
        
        # 2. Проходим по второму списку
        curr = headB
        while curr:
            # Если узел есть в первом списке — это точка пересечения
            if curr in nodes_in_a:
                return curr
            curr = curr.next
        return None
```

**Пояснение**: 
Благодаря хеш-таблице мы находим пересечение за один проход по второму списку после подготовки. Сложность $O(m + n)$ по времени.
