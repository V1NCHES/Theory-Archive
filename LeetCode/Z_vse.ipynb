1. Две суммы    https://leetcode.com/problems/two-sum/

Дан массив целых чисел nums и целое число target. Верните индексы этих двух чисел, сумма которых равнаtarget .

Можно предположить, что для каждого входного значения существует ровно одно решение , и нельзя использовать один и тот же элемент дважды.

Вы можете вернуть ответ в любом порядке.

Решение за O(n) с помощью хеш таблицы.

```python
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):# i - индекс, num - значения массив для этого индекса
            complemet = target - num #вычисляем разницу а дальше сморим находиться ли такое число в словаре, при это если она отрицательная занчит с таким число точно пану не связать
            if complemet in seen:# проверяем что если такое число в словаре, если нет то ищем дальше
                return [seen[complemet], i] # возрашаем ответ
            
            seen[num] = i
        
        return [] # если не нашли то возрашаем пустоту
    
if __name__ == "__main__":
    # Локальное тестирование
    solution = Solution()
    
    # Тест 1
    nums = [2, 7, 11, 15]
    target = 9
    result = solution.twoSum(nums, target)
    print(f"Тест 1: nums={nums}, target={target}")
    print(f"Результат: {result}")
    print(f"Проверка: {nums[result[0]]} + {nums[result[1]]} = {target}")
    
    print()
```

9. Число-палиндром   https://leetcode.com/problems/palindrome-number/

Если задано целое число x, верните значение, true если x это число. палиндроми falseв противном случае .

Желатено не использовать преобразование в строку.

решение мое, с выделение памяти: Среда выполнения 36 РС Биты 5,31 %Память19.40МББиты18,94 %

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if(x < 0):
            return False
        arr = []
        i = 1
        cur = int(x//10**i)
        arr.append(x - cur*10**i)
        while( cur  > 0):    
            i+=1
            cur =int( x//10**i)
            arr.append(int((x - cur*10**i)//10**(i-1)))
        l = len(arr)
        diapazon = l//2

        for i in range(0,diapazon,1):
            if arr[i] !=  arr[l-i -1]:
                return False

        return True
        
solution = Solution()
print(solution.isPalindrome(12231))
        
```

Самы оптимальный без выделения памяти, но есть проблема что долго рабоает все равно, но уже без выделения памяти, но можно пользоваться срезами и сделать через строку

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        # Отрицательные числа и числа, оканчивающиеся на 0 (кроме самого 0)
        if x < 0 or (x % 10 == 0 and x != 0):
            return False       
        reversed_half = 0
        while x > reversed_half:
            # Берём последнюю цифру x и добавляем к reversed_half
            reversed_half = reversed_half * 10 + x % 10
            x //= 10
        
        # Для чётной длины: x == reversed_half
        # Для нечётной длины: x == reversed_half // 10 (отбрасываем среднюю цифру)
        return x == reversed_half or x == reversed_half // 10

solution = Solution()
print(solution.isPalindrome(12231))
```

Римские цифры обозначаются семью различными символами:  I, V, X, L, C, Dи M.

Символ        Значение
Я 1

Например,  число 2записывается как II в римской транслитерации, просто путем сложения двух единиц. 12Число записывается как  XII, что просто означает X + II. Число 27записывается как XXVII, что означает XX + V + II.

Римские цифры обычно записываются от наибольшей к наименьшей слева направо. Однако цифра четыре записывается не так IIII. Вместо этого число четыре записывается как IV. Поскольку единица стоит перед пятеркой, мы вычитаем ее, получая четыре. Тот же принцип применим к числу девять, которое записывается как IX. Существует шесть случаев, когда используется вычитание:

Iможет быть размещен перед V(5) и X(10), чтобы получить 4 и 9. 
Xможет быть размещен перед L(50) и C(100), чтобы получить 40 и 90. 
Cможет быть размещен перед D(500) и M(1000), чтобы получить 400 и 900.
Дана римская цифра, преобразуйте её в целое число.

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        m = {'I':1, 'V':5, 'X':10, 'L':50, 'C':100, 'D':500, 'M':1000}
        return sum(m[s[i]]  if (i == len(s) - 1 or m[s[i]] >= m[s[i+1]]) else (-m[s[i]]) for i in range(len(s)))


solution = Solution()
print(solution.romanToInt("III"))
```

14. Самый длинный общий префикс https://leetcode.com/problems/longest-common-prefix/description/

Напишите функцию для поиска самой длинной общей префиксной строки среди массива строк.

Если общего префикса нет, вернуть пустую строку "".

```python
from typing import List  # ← ЭТО ВАЖНО!

class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        if not strs:  # Проверка на пустой массив
            return ""
        
        # Берём первую строку как эталон
        prefix = strs[0]
        
        # Сравниваем с каждой строкой
        for s in strs[1:]:
            # Укорачиваем prefix, пока он не станет префиксом s
            while not s.startswith(prefix):
                prefix = prefix[:-1]  # отрезаем последний символ
                if not prefix:
                    return ""
        
        return prefix

#["flower","flow","flight"];  ["dog", "racecar", "car"]
solution = Solution()
print(solution.longestCommonPrefix(["flower","flow","flight"]  ))  # Вывод: "" (нет общего префикса)
```

20. Допустимые скобки   https://leetcode.com/problems/valid-parentheses/

Дана строка, s  содержащая только символы '(', ')', '{', '}', '['и ']'. Определите, является ли входная строка допустимой.

Входная строка считается допустимой, если:

Открытые скобки должны быть закрыты скобками того же типа.
Открытые скобки необходимо закрыть в правильном порядке.
Каждой закрывающей скобке соответствует открывающая скобка того же типа.

```python
from collections import deque

class Solution:
    def isValid(self, s: str) -> bool:
        stack = deque()
        
        # Словарь для соответствия скобок (быстрее и чище)
        bracket_map = {')': '(', '}': '{', ']': '['}
        
        for char in s:
            if char in bracket_map:  # закрывающая скобка, так как храним в словаре, доступ за O(1)
                # Проверяем, пуст ли стек ИЛИ не совпадает ли верхушка
                if not stack or stack[-1] != bracket_map[char]:
                    return False
                stack.pop()
            else:  # открывающая скобка
                stack.append(char)
        
        return not stack  # True если стек пуст

solution = Solution()
print(solution.isValid("{}")) 
```

21. Объединение двух отсортированных списков   https://leetcode.com/problems/merge-two-sorted-lists/

Вам даны заголовки двух отсортированных связанных списков list1и list2.

Объедините два списка в один отсортированный список. Список должен быть создан путем объединения узлов первых двух списков.

Возвращает начало объединенного связанного списка .

```python
from typing import Optional

# 1. Определение ListNode (обязательно)
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    # Для красивого вывода списка
    def __repr__(self):
        if self.next:
            return f"{self.val} -> {self.next}"
        return str(self.val)

# 2. Решение
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode()
        current = dummy
        
        while list1 and list2:
            if list1.val <= list2.val:
                current.next = list1
                list1 = list1.next
            else:
                current.next = list2
                list2 = list2.next
            current = current.next
        
        current.next = list1 if list1 else list2
        
        return dummy.next

# 3. Вспомогательная функция для создания списков (удобно для тестов)
def create_list(arr):
    """Создаёт связный список из обычного списка Python"""
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

# 4. Тестирование
if __name__ == "__main__":
    # Создаём тестовые списки
    list1 = create_list([1, 2, 4])
    list2 = create_list([1, 3, 4])
    
    print("Список 1:", list1)
    print("Список 2:", list2)
    
    # Запускаем решение
    solution = Solution()
    result = solution.mergeTwoLists(list1, list2)
    
    print("Результат:", result)  # 1 -> 1 -> 2 -> 3 -> 4 -> 4
    
    # Дополнительные тесты
    print("\n--- Другие тесты ---")
    
    # Тест с пустыми списками
    print("Пустые списки:", Solution().mergeTwoLists(None, None))
    
    # Тест с одним пустым
    list3 = create_list([0])
    print("Один пустой:", Solution().mergeTwoLists(list3, None))
```

26. Удалите дубликаты из отсортированного массива. https://leetcode.com/problems/remove-duplicates-from-sorted-array/description/

Дан массив целых чисел , numsотсортированный в порядке неубывания . Необходимо удалить дубликаты на месте таким образом, чтобы каждый уникальный элемент встречался только один раз . Относительный порядок элементов должен оставаться неизменным .

Считайте количество уникальных элементов равным  nums. После удаления дубликатов верните количество уникальных элементов  .k​​​​​​​k

Первые  k элементы массива  nums должны содержать уникальные числа в отсортированном порядке . Остальные элементы после индекса  k - 1 можно игнорировать.

```python
from typing import List

class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        l = len(nums)
        k = l
        
        # Идём с конца, чтобы индексы не сбивались
        for i in range(l - 1, 0, -1):  # до 0 включительно (индекс 1 проверяем, сравнивая с 0)
            if nums[i] == nums[i - 1]:
                nums.pop(i)  # удаляем дубликат, а не последний элемент
                k -= 1
        
        print(nums)
        return k


# Вызов ВНЕ класса
solution = Solution()
result = solution.removeDuplicates([1, 1, 2])
print(f"Результат: {result}")  # [1, 2], k = 2
```
